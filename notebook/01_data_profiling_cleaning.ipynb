## IMPORT DATASETS

```python
import pandas as pd
employee_survey = pd.read_csv('../data/raw/employee_survey_data.csv')
general_data = pd.read_csv('../data/raw/general_data.csv')
manager_survey = pd.read_csv('../data/raw/manager_survey_data.csv')

# Column names of each dataset
print("--- EMPLOYEE SURVEY COLUMNS ---")
print(employee_survey.columns.tolist())

print("\n--- GENERAL DATA COLUMNS ---")
print(general_data.columns.tolist())

print("\n--- MANAGER SURVEY COLUMNS ---")
print(manager_survey.columns.tolist())
```

## general_data TABLE CLEANING

```python
# Data structure and first few rows
print("--- GENERAL DATA PREVIEW ---")
display(general_data.head())

# Column types and missing values
print("\n--- GENERAL DATA INFO ---")
general_data.info()

# Statistical summary
print("\n--- GENERAL DATA SUMMARY ---")
display(general_data.describe())
```

Above result shows NumCompaniesWorked and TotalWorkingYears columns have missing values.
Some columns are maybe useless for analysis.

### Cross check unique values for suspected useless columns

```python
print(f"Unique values in EmployeeCount: {general_data['EmployeeCount'].unique()}")
print(f"Unique values in Over18: {general_data['Over18'].unique()}")
print(f"Unique values in StandardHours: {general_data['StandardHours'].unique()}")
```

### Drop useless columns

```python
columns_to_drop = ['EmployeeCount', 'Over18', 'StandardHours']
general_data = general_data.drop(columns=columns_to_drop)
print("New shape of general_data:", general_data.shape)
```

### Handle missing values

```python
# Fill missing TotalWorkingYears using the median of the employee's exact Age
general_data['TotalWorkingYears'] = general_data.groupby('Age')['TotalWorkingYears'].transform(lambda x: x.fillna(x.median()))

# Fill missing NumCompaniesWorked using the median of the employee's exact Age
general_data['NumCompaniesWorked'] = general_data.groupby('Age')['NumCompaniesWorked'].transform(lambda x: x.fillna(x.median()))

# Check if there are any remaining missing values in general_data
print("Remaining missing values in general_data:")
print(general_data[['TotalWorkingYears', 'NumCompaniesWorked']].isnull().sum())
```

## employee_survey TABLE CLEANING

```python
print("--- Employee Survey PREVIEW ---")
display(employee_survey.head())

# Column types and missing values
print("\n--- Employee Survey INFO ---")
employee_survey.info()

# Value counts for the Employee Survey columns
print("--- EMPLOYEE SURVEY RATINGS BREAKDOWN ---")
for col in ['EnvironmentSatisfaction', 'JobSatisfaction', 'WorkLifeBalance']:
    print(f"\nValue Counts for {col}:")
    print(employee_survey[col].value_counts().sort_index())
```

Above result shows 3 columns have missing values
### Handle missing values

```python
# Fill missing survey values with their respective column medians
employee_survey['EnvironmentSatisfaction'] = employee_survey['EnvironmentSatisfaction'].fillna(employee_survey['EnvironmentSatisfaction'].median())
employee_survey['JobSatisfaction'] = employee_survey['JobSatisfaction'].fillna(employee_survey['JobSatisfaction'].median())
employee_survey['WorkLifeBalance'] = employee_survey['WorkLifeBalance'].fillna(employee_survey['WorkLifeBalance'].median())

# Verify that there are zero missing values left
print("Remaining missing values in employee_survey:")
print(employee_survey.isnull().sum())
```

## manager_survey TABLE

```python
# 1. See the actual data structure and first few rows
print("--- Manager Survey PREVIEW ---")
display(manager_survey.head())

# 2. Check column types and look for missing (null) values
print("\n--- MANAGER Survey INFO ---")
manager_survey.info()

# Check the value counts for the Manager Survey columns
print("\n--- MANAGER SURVEY RATINGS BREAKDOWN ---")
for col in ['JobInvolvement', 'PerformanceRating']:
    print(f"\nValue Counts for {col}:")
    print(manager_survey[col].value_counts().sort_index())
```

This table is already clean
## MERGE TABLES

```python
# Merge all three cleaned dataframes together on EmployeeID
merged_df = general_data.merge(employee_survey, on='EmployeeID', how='inner')
final_df = merged_df.merge(manager_survey, on='EmployeeID', how='inner')

expected_rows = 4410
actual_rows = final_df.shape[0]
duplicate_ids = final_df.duplicated(subset=['EmployeeID']).sum()

# Validation check
assert final_df.shape[0] == 4410, f"Wrong row count: {final_df.shape[0]}"
assert final_df['EmployeeID'].is_unique, "Duplicate employees found!"
assert final_df.shape[0] == general_data.shape[0], "Row multiplication detected!"

print("All checks passed perfectly!")
```

## SAVE THE FILE

```python
final_df.to_csv('../data/processed/hr_employee_master.csv', index=False)
print("\nFile saved successfully as 'hr_employee_master.csv'.")
```
